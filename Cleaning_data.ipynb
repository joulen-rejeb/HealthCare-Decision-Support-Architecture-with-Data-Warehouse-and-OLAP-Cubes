```python
import pandas as pd
import numpy as np
from scipy import stats
```

```python
file_path = 'C:\\Users\\man\\Downloads\\base-contrat-complete.xlsm'
df = pd.read_excel(file_path, sheet_name='Base Complète')
df.head()
```

```python
# Check for missing values in each column
missing_values = df.isnull().sum()

# Print the missing values count for each column
print(missing_values)

# Remove rows with missing values (if necessary)
df_cleaned = df.dropna()

# Alternatively, replace missing values with the mean (for numerical columns)
# df.fillna(df.mean(), inplace=True)
```

```python
# Step 1: Handle Outliers Using the IQR Method for specific columns

# List of columns to handle outliers
columns_to_check = ['Prime_brute', 'Prime_nette', 'Prime Annuelle', 'tot_prime_nette']

# Select only the specified numerical columns for outlier detection
df_selected = df_cleaned[columns_to_check]

# Calculate the IQR (Interquartile Range) for each column in the selected list
Q1 = df_selected.quantile(0.25)
Q3 = df_selected.quantile(0.75)
IQR = Q3 - Q1

# Define the lower and upper bounds for filtering out outliers for each column
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Create a mask to identify rows with outliers in the selected columns
mask_no_outliers = ~((df_selected < lower_bound) | (df_selected > upper_bound)).any(axis=1)

# Filter the original DataFrame to exclude rows with outliers in the specified columns
df_no_outliers_selected = df_cleaned[mask_no_outliers]


# Check the number of rows after removing the outliers
num_rows_after_outliers = df_no_outliers_selected.shape[0]
print(f"The number of rows after handling outliers: {num_rows_after_outliers}")
```

```python
# Remove duplicate rows
df_no_duplicates = df_no_outliers_selected.drop_duplicates()

# Display the cleaned data after removing duplicates
print("Data after removing duplicates:")
df_no_duplicates.head()
```

```python
# Get the number of rows after cleaning outliers and duplicates
num_rows_after_cleaning = df_no_duplicates.shape[0]
print(f"The number of rows after cleaning: {num_rows_after_cleaning}")
```

```python
# Save the cleaned data to the specified directory
df_no_duplicates.to_excel(r'C:\Users\man\Downloads\cleaned_data_final.xlsx', index=False)
```
